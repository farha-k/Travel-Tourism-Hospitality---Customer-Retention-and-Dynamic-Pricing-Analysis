```python
import pandas as pd
df = pd.read_csv('hotel_bookings.csv')
df.head()
```

```python
df.info()
df.isnull().sum()
```

```python
# Fill missing values

df['children'].fillna(0, inplace=True)
df['country'].fillna('Unknown', inplace=True)
df['agent'].fillna(0, inplace=True)
df['company'].fillna(0, inplace=True)

# Remove extreme outliers in price
df = df[df['adr'] < 500]

# Check again
df.isnull().sum()
```

```python
cancel_rate = df['is_canceled'].mean()
print("Cancellation Rate:", cancel_rate)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.boxplot(x='is_canceled', y='lead_time', data=df)
plt.title('Cancellation vs Lead Time')
plt.show()
```

```python
sns.countplot(x='deposit_type', hue='is_canceled', data=df)
plt.title("Cancellation by Deposit Type")
plt.xticks(rotation=30)
plt.show()
```

```python
sns.countplot(x='market_segment', hue='is_canceled', data=df)
plt.title("Cancellation by Market Segment")
plt.xticks(rotation=45)
plt.show()
```

```python
sns.boxplot(x='is_canceled', y='adr', data=df)
plt.title("Cancellation vs Price (ADR)")
plt.show()
```
