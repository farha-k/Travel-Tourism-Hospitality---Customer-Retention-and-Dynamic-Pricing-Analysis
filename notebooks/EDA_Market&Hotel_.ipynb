```
# Week 2: Exploratory Data Analysis (EDA)
```

```
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```
df = pd.read_csv('/content/hotel_bookings.csv')
df.head()
```

```
df.info()
df.shape
```

```
df.isnull().sum()

df['children'].fillna(0, inplace=True)
df['country'].fillna('Unknown', inplace=True)
df['agent'].fillna(0, inplace=True)
df['company'].fillna(0, inplace=True)
```

```
sns.countplot(x='hotel', data=df)
plt.title("Hotel Type Count")
plt.show()
```

```
sns.countplot(x='is_canceled', data=df)
plt.title("Cancellation Count")
plt.show()
```

```
top_countries = df['country'].value_counts().head(5)
sns.barplot(x=top_countries.values, y=top_countries.index)
plt.title("Top 5 Countries")
plt.show()
```

```
sns.histplot(df['lead_time'], bins=50)
plt.title("Lead Time Distribution")
plt.show()
```

```
sns.countplot(x='hotel', hue='is_canceled', data=df)
plt.title("Cancellation by Hotel Type")
plt.show()
```

# Insights

- City Hotels have more bookings compared to Resort Hotels.
- Cancellation count is high (~44k bookings canceled).
- Most customers are from Portugal.
- Lead time is high, meaning customers book early.

# Conclusion

EDA helped understand booking patterns, cancellations, and customer distribution.
This analysis supports better decision-making in pricing and retention strategies.
